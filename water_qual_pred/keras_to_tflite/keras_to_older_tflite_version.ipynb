```python
import tensorflow as tf
import os

# 1. Paths (Update these for your local setup or Colab)
keras_model_path = 'water_qual_param.keras'
tflite_output_path = 'water_qual_flutter.tflite'

if os.path.exists(keras_model_path):
    print("🧠 Loading Master Keras Model...")
    model = tf.keras.models.load_model(keras_model_path)

    # 2. Setup Converter with Flutter/Mobile Optimization
    converter = tf.lite.TFLiteConverter.from_keras_model(model)

    # --- ENABLING FLEX DELEGATE (CRITICAL FOR LSTMs) ---
    # Standard TFLite doesn't have all LSTM math; SELECT_TF_OPS pulls them in.
    converter.target_spec.supported_ops = [
        tf.lite.OpsSet.TFLITE_BUILTINS,
        tf.lite.OpsSet.SELECT_TF_OPS
    ]

    # Fix for 'tf.TensorListReserve' which often crashes Flutter LSTMs
    converter._experimental_lower_tensor_list_ops = False

    # Optimize for latency and size (standard for mobile)
    converter.optimizations = [tf.lite.Optimize.DEFAULT]

    # 3. Execution
    print("🚀 Converting... Please wait.")
    tflite_model = converter.convert()

    # 4. Save the file
    with open(tflite_output_path, 'wb') as f:
        f.write(tflite_model)

    print(f"✅ SUCCESS! Model saved as: {tflite_output_path}")

    # 5. Print Input/Output details for your Flutter Teammates
    interpreter = tf.lite.Interpreter(model_content=tflite_model)
    interpreter.allocate_tensors()
    print("\n📝 FLUTTER INTEGRATION INFO:")
    print(f"Input Shape: {interpreter.get_input_details()[0]['shape']}")
    print(f"Output Shape: {interpreter.get_output_details()[0]['shape']}")
else:
    print(f"❌ Error: {keras_model_path} not found!")
```

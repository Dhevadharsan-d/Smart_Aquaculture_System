```python
from google.colab import drive
from ultralytics import YOLO
import os


# 1. Mount Drive (This connects your Google Drive to Colab)
drive.mount('/content/drive')
```

```python

import cv2


ROOT_FOLDER = '/content/drive/MyDrive/smart_aquaculture/yolov8_project/Third_update_testing'

# 1. Load your newly retrained 'best.pt'
model = YOLO(f'{ROOT_FOLDER}/third_finalized_models/third_yolo.pt')

# 2. Open your test video


# output_dir = f'{ROOT_FOLDER}/third_video_analytics_output'
OUTPUT_FOLDER = f'{ROOT_FOLDER}/third_video_analytics_output'

if not os.path.exists(OUTPUT_FOLDER):
    os.makedirs(OUTPUT_FOLDER)
    print(f"Created new folder: {OUTPUT_FOLDER}")

output_video_path = f'{OUTPUT_FOLDER}/tracking_result_v3.mp4'

video_path = f'{ROOT_FOLDER}/third_testing_video/video3.mp4'
cap = cv2.VideoCapture(video_path)
```

```python
# Get video details
w, h, fps = (int(cap.get(x)) for x in (cv2.CAP_PROP_FRAME_WIDTH, cv2.CAP_PROP_FRAME_HEIGHT, cv2.CAP_PROP_FPS))
out = cv2.VideoWriter(output_video_path, cv2.VideoWriter_fourcc(*'mp4v'), fps, (w, h))
```

```python
print(f"Processing video... 🚀 Saving to: {output_video_path}")

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    # Use .track() for temporal stability as part of your IoT monitoring method [1]
    # persist=True maintains object IDs across frames
    results = model.track(frame, persist=True, imgsz=640, conf=0.3)

    # Draw boxes and IDs
    annotated_frame = results[0].plot()

    # Write the frame to the file in your Drive folder
    out.write(annotated_frame)

cap.release()
out.release()
print(f"✅ Done! Video saved successfully in: {OUTPUT_FOLDER}")
```
